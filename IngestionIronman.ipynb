```python
import fitz

pdf_path = "/content/drive/MyDrive/ironman1script.pdf"

def extract_text(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""
    for page in doc:
        text += page.get_text()
    return text

script_text = extract_text(pdf_path)

print(script_text[:1000])  # sanity check
```

```python
import re

def clean_script(text):
    # Remove copyright blocks
    text = re.sub(r"©.*?\n", "", text)
    text = re.sub(r"NO DUPLICATION.*?\n", "", text)

    # Remove revision color markers
    text = re.sub(r"SALMON.*?\n", "", text)
    text = re.sub(r"GOLDENROD.*?\n", "", text)
    text = re.sub(r"GREEN.*?\n", "", text)
    text = re.sub(r"YELLOW.*?\n", "", text)
    text = re.sub(r"PINK.*?\n", "", text)
    text = re.sub(r"BLUE.*?\n", "", text)
    text = re.sub(r"WHITE.*?\n", "", text)
    text = re.sub(r"BUFF.*?\n", "", text)
    text = re.sub(r"XX/XX/07", "", text)

    # Remove page numbers
    text = re.sub(r"\n\d+\.\n", "\n", text)
    text = re.sub(r"\n\d+\n", "\n", text)

    # Remove continuation markers
    text = re.sub(r"\(CONTINUED.*?\)", "", text)
    text = re.sub(r"\(MORE\)", "", text)

    return text

cleaned_text = clean_script(script_text)
```

```python
print(cleaned_text[:1000])
```

```python
def split_scenes(text):
    pattern = r"\n(?=(INT\.|EXT\.))"
    chunks = re.split(pattern, text)

    scenes = []
    current_scene = ""

    for chunk in chunks:
        if chunk in ["INT.", "EXT."]:
            if current_scene:
                scenes.append(current_scene.strip())
            current_scene = chunk
        else:
            current_scene += chunk

    if current_scene:
        scenes.append(current_scene.strip())

    return scenes

scenes = split_scenes(cleaned_text)
print("Total scenes:", len(scenes))
print(scenes[0][:400])
```

```python
print(scenes[9])
```

```python
from langchain_core.documents import Document

def extract_characters(scene_text):
    characters = []
    possible = ["TONY", "RHODEY", "STANE", "PEPPER", "YINSEN"]
    for name in possible:
        if name in scene_text:
            characters.append(name)
    return characters

documents = []

for i, scene in enumerate(scenes):
    lines = scene.split("\n")
    heading = lines[0]

    scene_type = "UNKNOWN"
    location = "UNKNOWN"
    time_of_day = "UNKNOWN"

    if heading.startswith("INT.") or heading.startswith("EXT."):
        scene_type = heading.split(".")[0]
        parts = heading.split(" - ")
        if len(parts) >= 2:
            location = parts[0].replace("INT.", "").replace("EXT.", "").strip()
            time_of_day = parts[-1].strip()

    documents.append(
        Document(
            page_content=scene,
            metadata={
                "scene_number": i + 1,
                "scene_type": scene_type,
                "location": location,
                "time_of_day": time_of_day,
                "characters": extract_characters(scene),
                "movie": "Iron Man (2008)"
            }
        )
    )

print("Documents ready:", len(documents))
```

```python
def trim_to_first_scene(text):
    match = re.search(r"\n(INT\.|EXT\.)", text)
    if match:
        return text[match.start():]
    return text
```

```python
cleaned_text = clean_script(script_text)
cleaned_text = trim_to_first_scene(cleaned_text)

print(cleaned_text[:500])
```

```python
def split_scenes(text):
    pattern = r"\n(?=INT\.|EXT\.)"
    scenes = re.split(pattern, text)

    # Keep only valid scenes
    scenes = [s.strip() for s in scenes if s.strip().startswith(("INT.", "EXT."))]

    return scenes
```

```python
scenes = split_scenes(cleaned_text)

print("Total scenes:", len(scenes))
print(scenes[0][:400])
```

```python
from langchain_core.documents import Document

documents = []

for i, scene in enumerate(scenes):
    lines = scene.split("\n")
    heading = lines[0]

    scene_type = heading.split(".")[0]

    parts = heading.split(" - ")
    location = parts[0].replace("INT.", "").replace("EXT.", "").strip()
    time_of_day = parts[-1].strip() if len(parts) > 1 else "UNKNOWN"

    documents.append(
        Document(
            page_content=scene,
            metadata={
                "scene_number": i + 1,
                "scene_type": scene_type,
                "location": location,
                "time_of_day": time_of_day,
                "movie": "Iron Man (2008)"
            }
        )
    )

print("Documents ready:", len(documents))
```

```python
vector_store.add_documents(documents)
```

```python
results = vector_store.similarity_search(
    "Tony builds the first suit in the cave",
    k=2
)

for r in results:
    print(r.metadata)
    print(r.page_content[:300])
    print("------")
```
